```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings

warnings.filterwarnings("ignore")
```

```python
from sklearn.datasets import load_iris

iris = load_iris()
iris
```

```python
X = iris.data
y = iris.target
```

```python
df = pd.DataFrame(
    X,
    columns=iris.feature_names
)
df['target'] = y
df.head()
```

```python
from sklearn.model_selection import train_test_split

X_train,X_test,y_train,y_test = train_test_split(
    X,y,test_size=0.3,random_state=42
)
```

```python
from sklearn.tree import DecisionTreeClassifier

tree_classifier = DecisionTreeClassifier()
```

```python
tree_classifier.fit(X_train,y_train)
```

```python
y_pred = tree_classifier.predict(X_test)
```

```python
# Visualize Decision tree
from sklearn import tree

plt.figure(figsize=(15,10))
tree.plot_tree(tree_classifier,filled=True)
plt.show()
```

```python
from sklearn.metrics import confusion_matrix,accuracy_score,classification_report

print("Confusion matrix", confusion_matrix(y_test,y_pred))
print("Accuracy:", accuracy_score(y_test,y_pred))
print("Classification report:", classification_report(y_test,y_pred))
```

# Post Pruning using cost complexity pruning

```python
path = tree_classifier.cost_complexity_pruning_path(X_train, y_train)
ccp_alphas = path.ccp_alphas
ccp_alphas
```

# different models for different ccp_alpha values

```python
models = []

for alpha in ccp_alphas:
    model = DecisionTreeClassifier(
        ccp_alpha=alpha,
        random_state=42
    )
    model.fit(X_train, y_train)
    models.append(model)
```

```python
models
```

```python
train_scores = [model.score(X_train, y_train) for model in models]
test_scores = [model.score(X_test, y_test) for model in models]
```

```python
train_scores, test_scores
```

```python
# Visualizing ccp_alpha and accuracy
import matplotlib.pyplot as plt

plt.plot(ccp_alphas, train_scores,marker='o', label='Train')
plt.plot(ccp_alphas, test_scores,marker='o', label='Test')
plt.xlabel('CCP Alpha')
plt.ylabel('Accuracy')
plt.legend()
plt.show()
```

# Decision Tree Classifier PrePruning and Hyperparameter tuning

```python
param = {
    'criterion': ['gini','entropy','log_loss'],
    'splitter': ['best', 'random'],
    'max_depth': [2,3,4,5,6,7],
    'min_samples_split': [3,4,5,6],
    'min_samples_leaf': [1,2,3,4,5],
    'max_features': ['sqrt', 'log2']
}
```

```python
from sklearn.model_selection import GridSearchCV

grid = GridSearchCV(
    estimator=tree_classifier,
    param_grid=param,
    cv=5,
    scoring='accuracy'
)

grid.fit(X_train,y_train)
```

```python
print("Best Parameters:", grid.best_params_)
print("Best Accuracy:", grid.best_score_)
best_model = grid.best_estimator_
```

```python
y_pred = best_model.predict(X_test)
```

```python
print("Confusion matrix", confusion_matrix(y_test,y_pred))
print("Accuracy:", accuracy_score(y_test,y_pred))
print("Classification report:", classification_report(y_test,y_pred))
```
