# PCA Implementation

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings

warnings.filterwarnings('ignore')
```

```python
# Load dataset
from sklearn.datasets import load_breast_cancer
```

```python
cancer_df = load_breast_cancer()
```

```python
print(cancer_df.DESCR)
```

```python
df = pd.DataFrame(cancer_df.data,columns=cancer_df.feature_names)
df.head()
```

```python
y = cancer_df.target
```

```python
# Standardization
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
scaler.fit(df)
```

```python
scaled_data = scaler.transform(df)
```

```python
scaled_data
```

```python
# PCA
from sklearn.decomposition import PCA

pca = PCA(n_components=2)
data_pca = pca.fit_transform(scaled_data)
```

```python
data_pca
```

```python
# Explained Variance
print("Explained variance:", pca.explained_variance_ratio_)
print("Total variance:", sum(pca.explained_variance_ratio_))
```

```python
# Visualizing data after applying PCA

plt.figure(figsize=(8, 6))

plt.scatter(data_pca[:,0],data_pca[:,1],c=y)
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("PCA Visualization")
plt.legend()
plt.grid()
plt.show()
```
